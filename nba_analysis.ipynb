```python
import pandas as pd

df = pd.read_csv('NBA Stats 202324 All Stats  NBA Player Props Tool (4).csv')
print(df.shape)
df.head()

# Check column names and basic stats 
print(df.columns.tolist())
print(df.describe())

```

```python
# Top 10 scorers 
top_scorers = df[['NAME', 'TEAM', 'POS', 'PPG']].sort_values('PPG', ascending=False).head(10)
print(top_scorers)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.barh(top_scorers['NAME'], top_scorers['PPG'], color='navy')
plt.xlabel('Points Per Game')
plt.title('Top 10 NBA Scorers 2023-24 Season')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.savefig('top_scorers.png')
plt.show()
```

```python
# Age vs Scoring - does age affect PPG? 
plt.figure(figsize=(10, 6))
plt.scatter(df['AGE'], df['PPG'], color='navy', alpha=0.6)
plt.xlabel('Age')
plt.ylabel('Points Per Game')
plt.title('Age vs Scoring - NBA 2023-24 Season')
plt.tight_layout()
plt.savefig('age_vs_scoring.png')
plt.show()
```

```python
import numpy as np 

# Add trend line 
z = np.polyfit(df['AGE'], df['PPG'], 1)
p = np.poly1d(z)

plt.figure9figsize=(10, 6)
plt.scatter(df['AGE'], df['PPG'], color='navy', alpha=0.6)
plt.plot(sorted(df['AGE']), p(sorted(df['AGE'])), color='red', linewidth=2)
plt.xlabel('Age')
plt.ylabel('Points Per Game')
plt.title('Age vs Scoring - NBA 2023-24 Season')
plt.tight_layout()
plt.savefig('age_vs_scoring_trend,png')
plt.show()
```

```python
# Average PPG by position
avg_by_position = df.groupby('POS')['PPG'].mean().sort_values(ascending=False)

plt.figure(figsize=(10, 6))
plt.bar(avg_by_position.index, avg_by_position.values, color='navy')
plt.xlabel('Position')
plt.ylabel('Average Points Per Game')
plt.title('Average Scoring by Position - NBA 2023-24 Season')
plt.tight_layout()
plt.savefig('scoring_by_position.png')
plt.show()
```
